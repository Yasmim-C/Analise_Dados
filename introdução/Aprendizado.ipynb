# Aprendendo python   

## Verificar quais bibliotecas tem instlado
pip list 

## Verifcar a versão da biblioteca
pip freeze

## Informações da biblioteca
pip show <nome da biblioteca> / pip show matplotlib

## Incluindo as bibliotecas instaladas no projeto dentro do requirements.txt. E sempre que uma nova biblioteca for incluída, rodar o pip novamente.
pip freeze > requirements.txt

![image.png](../images/requirements.png)

## Módulo nativo Venv. Configurando a variavel de ambiente

python -m venv .venv

### Instalando as dependencias de modo recursivo 


pip install -r requirements.txt

## Versionamento de projeto com Git e GitHub

download do git: https://git-scm.com/

ir no next até finalizar

## Após instalar rodar os comandos de configurar nome e endereço de email
abra o git bash no computador

git config --global user.name "Insira seu nome"
git config --global user.email email@email.com

Ir no CMD > Win + R e digitar CMD e escrever o comando git. Para validar que o git está instalado.

### Dentro do projeto, quando quiser que ele seja um repostorio vazio do git, executar o comando
git init

### Para verificar o status de quais alterações não foram commitadas utilizar o comando 
git status
