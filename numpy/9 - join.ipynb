# JOIN


```python
import numpy as np
```

```python
# Cirando dois arrays 1D

array1 =np.array([1, 2, 3,])
array2 =np.array([4, 5, 6])
```

```python
# Concatenando os arrays usando np.concatenate()
novo_array = np.concatenate((array1, array2))
novo_array
```

```python
# Criando arrays para nomes e notas

nomes =np.array(['Alice', 'Bob', 'Charlie'])
notas_quimica = np.array([85, 90, 78])
notas_fisica = np.array([80, 85, 70])
```

```python
# Usando np.vstack() para empilhar verticalmente os arrays de notas
notas_estudantes = np.vstack((nomes, notas_quimica, notas_fisica))
notas_estudantes
```

```python
# Usando np.hstack() para empilhar horizontalmente os arrays de notas
matriz_notas = np.column_stack((notas_quimica, notas_fisica))
matriz_notas
```

```python
# Adicionando uma nova coluna com as médias de notas

media_notas = np.mean(matriz_notas, axis=1)
matriz_final = np.column_stack((matriz_notas, media_notas))
matriz_final
```
