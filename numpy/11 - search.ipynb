# PESQUISANDO EM ARRAYS

```python
import numpy as np
```

```python
array = np.array([10, 20, 30, 40, 50])
array
```

```python
# Pesquisando valor 30 no array
posicao = np.where(array==30)
posicao
```

```python
# Criar um array com notas de alunos  
notas = np.array([7.5, 8.0, 6.5, 9.0, 5.5, 7.0, 9.0, 10.0])
notas
```

```python
# Pesquisar quais alunos foram aprovados (nota maior que 7)
alunos_aprovados = np.where(notas > 7)
notas[alunos_aprovados]
```

```python
nota_maxima = np.max(notas)
nota_minima = np.min(notas)
nota_maxima, nota_minima
```

```python
indice_maximo = np.argmax(notas)
indice_maximo
```
