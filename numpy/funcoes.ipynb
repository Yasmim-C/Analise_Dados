# Funções universais do NUMPY (uFunc)

```python
import numpy as np
```

```python
array = np.array([1, 2, 3, 4, 5])
array
```

```python
# Adição
soma = np.add(array, 10)
soma
```

```python
# Subtração
subtracao = np.subtract(array, 5)
subtracao
```

```python
# Multiplicação
produto = np.multiply(array, 2)
produto
```

```python
# Divisão
divisao = np.divide(array, 2)
divisao
```

```python
# Funções trigonométricas
#Seno
seno = np.sin(array)
seno
```

```python
# Cosseno
cosseno = np.cos(array)
cosseno
```

```python
# Funções estatisticas
# Média
media = np.mean(array)
media
```

```python
# Desvio Padrão
desvio_padrao = np.std(array)
desvio_padrao
```

```python
# Funções de comparação 
comparacao_maior_3 = array > 3
comparacao_maior_3
```

```python
# Combinando funções
resultado = np.sqrt(array) + np.power(array, 2)
resultado
```

```python
# Funções expoenciais
exponencial = np.exp(array)
exponencial
```

```python
# # Função de logaritmo
logaritmo = np.log(array)
logaritmo
```

```python
# Funções de arredondamento
arredondado = np.round(array / 2)
arredondado
```

```python
# Função de potência
potencia = np.power(array, 3)
potencia
```
