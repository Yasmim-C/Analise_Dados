# SPLIT

```python
import numpy as np
```

```python
array = np.array([[1, 2, 3, 4, 5, 6]])
array
```

```python
array_dividido = np.split(array, 3, axis=1)
array_dividido
```

```python
for i, value in enumerate(array_dividido):
    print(f"Parte {i+1}: -> {value}")
```

```python
# Criar um array 1D com temperaturas diarias de um mês(30 dias)
temperaturas = np.array([30, 32, 31, 29, 28, 27, 26, 
                         25, 24, 23, 22, 21,
                         20, 19, 18, 17, 16,
                         15, 19, 22, 32, 31,
                         19, 28, 27, 16, 25,
                         24, 33, 32, 31, 36])
temperaturas
```

```python
# Dividir as temperaturas em 4 semanas ( cada semana com 7 dias)
semanas = np.split(temperaturas, 4)
semanas
```

```python
# Temperaturas durante os dias da semana
for i, value in enumerate(semanas):
    print(f"Semana {i+1}: -> {value}")
```

```python
# media semanal das temperaturas
media_semanal = np.mean(semanas, axis=1)
media_semanal
```

```python
# Média das temperaturas da semana
for i, value in enumerate(media_semanal):
    print(f"Média das temperaturas da Semana {i+1}: -> {value:.2f}ºC")
```
