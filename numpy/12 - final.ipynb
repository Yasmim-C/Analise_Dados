# Consolidando os Conhecimentos


```python
import numpy as np
```

```python
# Criando dados de exemplo

nomes = np.array(['Alice', 'Bob', 'Charlie', 'David', 'Eva'])
notas_quimica = np.array([7, 8, 8.7, 9.5, 10])
notas_fisica = np.array([8, 7.5, 9, 6.5, 10])
```

```python
# Calculando a media das notas de química e física para cada aluno
media = np.mean(np.column_stack((notas_quimica, notas_fisica)), axis=1)
media
```

```python
# Identificando os alunos aprovados
alunos_aprovados = np.where(media >= 7.7)
nomes[alunos_aprovados]
```

```python
# Organizando os dados em uma tabela
dados_tabela = np.column_stack((nomes, notas_quimica, notas_fisica, media))
dados_tabela
```

```python
for i in alunos_aprovados[0]:
    print(f"Aluno: {dados_tabela[i][0]}, Média: {media[i]:.2f}, Status: Aprovado")
```

```python
# Dividir os alunos em grupos
grupos = np.array_split(nomes, 3)

for i, value in enumerate(grupos):
    print(f"Grupo {i +1}: --> {value}")
```
