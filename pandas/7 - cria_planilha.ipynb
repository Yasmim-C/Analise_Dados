```python
import pandas as pd
```

```python
dados = {
    'Time' : ['Flamengo', 'São Paulo', 'Palmeiras', 'Santos', 'Corinthians'],
    'Partidas' : [10, 10, 10, 10, 10],
    'Vitorias': [7, 6, 5, 4, 3],
    'Empates': [2, 1, 3, 3, 2],
    'Derrotas': [1, 3, 2, 3, 5],
    'Gols_Pro': [22, 16, 18, 14, 10],
    'Gols_Contra': [10, 15, 12, 16, 20],
}
df = pd.DataFrame(dados)
df
```

```python
df['Saldo_Gols'] = df['Gols_Pro'] - df['Gols_Contra']
df
```

```python
df['Pontos'] = df['Vitorias'] * 3 + df['Empates'] * 1
df
```

```python
df = df.sort_values(by='Pontos', ascending=False)
df
```

```python
df.to_excel("dados/campeonato_brasileiro.xlsx", index=False)
```
