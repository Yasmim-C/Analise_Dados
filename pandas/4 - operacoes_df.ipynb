```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Nome': ['Alice', 'Bob', 'Charlie', 'David', 'Emily'],
    'Idade': [25, 30, 37, 40, 28],
    'Cargo': ['Analista', 'Gerente', 'Ceo', 'Analista', 'Coordenador'],
    'Salario': [5000, 8000, 15000, 4800, 6400]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
# Busca os primeiros registros do DataFrame, e quando colocar um numero entre os parênteses ele busca aquelas linha em especifico, lembrando que a indexação começa sempre em 0.
df.head(3)
```

```python
# Busca os últimos registros do DataFrame
df.tail(2)
```

```python
# Mostra informações do DataFrame.
df.info()
```

```python
# Tras descição do dataFrame. Tras a qtd de registros, a média de cada um deles, o desvio padrão. O primeiro, segundo e 3 quartil. Os valores mínimo, máximo
df.describe()
```

```python
# Filtrar para mostrar salários maiores que > 5000
df[df['Salario'] > 5000]
df
```

```python
# Filtrar da menor para maior idade. Usando o "True" filtrar do menor para o maior. E o "False" do maior para o menor.
df.sort_values(by='Idade', ascending=False)
```

```python
df['Bônus'] = df['Salario'] * 0.1
df
```

```python
# Média de salário por cargo
df.groupby('Cargo')['Salario'].mean()
```

```python
# Gráfico com os salários
df.plot(
    kind= 'bar',
    x= 'Nome',
    y= 'Salario',
    title='Salário dos Funcionários',
    rot=45
)
```
