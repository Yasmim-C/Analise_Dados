# Utilizando o DataFrame

```python
import pandas as pd
```

```python
dados = {
    'Nome' : ['Alice', 'Bob', 'Charile'],
    'Idade' : [ 24, 30, 22],
    'Cidade' : ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte']
}
```

```python
df = pd.DataFrame(dados)
df
```

```python
# o .head trás as 5 primeiras linhas de um dataframe, se quiser puxar alguma linha especifica. só colocar o numero entre parenteses
df.head()
```

```python
# .tail puxa as 5 ultimas linhas do dataFrame
df.tail(1)
```

```python
dados_notas = {
    'Nome' : ['Alice', 'Charlie', 'David', 'Eva', 'Bob'],
    'Matemática': [8.0, 8.5, 7.5, 9.0, 10.0],
    'Física' : [7.5, 8.0, 6.7, 8.5, 9.5],
    'Química' : [8.5, 7.5, 8.0, 9.0, 10.0]
    }
dados_notas
```

```python
df_notas = pd.DataFrame(dados_notas)
df_notas
```

```python
# Calculando a média dos alunos
df_notas['Média'] = df_notas[['Matemática', 'Física', 'Química']].mean(axis=1)
df_notas
```

```python
# Identificar alunos aprovados com nota maior que 8.5
alunos_aprovados = df_notas[df_notas['Média'] >= 8.5]
alunos_aprovados
```

```python
# Exibir a média e os alunos aprovados
df_notas[['Nome', 'Média']]
```

```python
# Exbir alunos aprovados mostrado só a média e o nome

alunos_aprovados[['Nome', 'Média']]
```
